```python
#standardizations
import numpy as np

data = np.array([1,2,3,4,5,6,7])

mean = np.mean(data)
print("Mean:", mean)

std = np.std(data)
print("Standard Deviation:", std)

z_scores = (data - mean) / std

print("\nStandardized Values (Z-scores):")
for i in range(len(data)):
    print("Value:", data[i], " Z-score:", z_scores[i])
```
